```python
import numpy as np
import pandas as pd
import sklearn.datasets
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
print(dataset)
```

```python
df.head()
```

```python
df.shape
```

```python
x = df
y = dataset.target
```

```python
print(x)
```

```python
print(y)
```

### splitting data into traing data and test data

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=3)
```

```python
print(x.shape,x_train.shape,x_test.shape)
```

### Standardize the data 

```python
# prints standar deviation
# if our data has all the values in same range then standard deviation must be 1
print(dataset.data.std())
```

```python
scalar = StandardScaler()
```

```python
scalar.fit(x_train)
```

```python
# now we will tranform the data based on this standard scalar
x_train_standardize = scalar.transform(x_train)
```

```python
x_train_standardize
```

```python
x_test_standardize = scalar.transform(x_test)
```

```python
print(x_train_standardize.std())
```

```python
print(x_test_standardize.std())
```

```python
scalar = StandardScaler()
```

We should never perform standardization before train-test split function because our ML model would directly have access to the test data before even performing the train-test split. So it is not a good and a practical thing to do. So for a good working ML model, we should always perform train-test split function first and then standardize the X-train function only so that we can perform our ML model with more accuracy.

```python
scalar.fit(x)
standardized_data = scalar.transform(x)
print(standardized_data)
```

# HANDLE IMBALANCE DATASET

```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('credit_data.csv')
```

```python
data.head()
```

```python
data.tail()
```

```python
data['Class'].value_counts()
```

This is Highly imbalance dataset 
                                                 
0 --> Legit Transactions                          
1 --> Fraudulent Transaction

```python
# sepearatign data
legit = data[data.Class == 0]
fraud = data[data.Class == 1]
print(legit.shape)
print(fraud.shape)
```

### UnderSampling

We are just building sample dataset containing similar distribution of legit and fraudelent Transaction                               
Number of Fadulent transaction : 492


```python
legit_sample = legit.sample(n = 492)
legit_sample.shape
```

Now , Concatinate the 2 data frames

```python
# axis = 0 means we want to concate the data frame one top of the other
new_dataset = pd.concat([legit_sample,fraud],axis = 0)
```

```python
new_dataset.head()
```

```python
new_dataset.tail()
```

```python
new_dataset['Class'].value_counts()
```
